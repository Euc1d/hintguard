```python
# Try your own tutor answer
import json
from hintguard import core
task = next(json.loads(l) for l in open('data/mbpp.jsonl') if json.loads(l)['task_id'] == 66)
answer = '''Use a counter and loop over the list; add one when the number is >= 0.'''
print(core.check(answer, task['code'].replace('\r\n', '\n'), task['test_list']))
```
